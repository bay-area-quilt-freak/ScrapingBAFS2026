```python
# solution for finding the average salary
police_names = ['Lt. Timothy Dolan', 'Sgt. Francisco Javier Negrete', 'Battalion Chief Christopher A Foley']
salaries = [711319.04, 666445.09, 589161.52]

total_salary = 0
for index, cop in enumerate(police_names):
    total_salary += salaries[index]

# you know there are 3 cops in the list
average_salary = total_salary / 3
print(average_salary)
```

```python
# or we can make the formating for the print statements fancy!!!
total_salary = 0
for index, cop in enumerate(police_names):
    print(f"{index + 1}. {cop}: ${salaries[index]:,.2f}")
    total_salary += salaries[index]

# you know the list has three items but if you didn't
# you can use len() to find the length of the list!!!
average_score = total_salary / len(police_names)

#let me know if you want me to breakdown this formating
print(f"Average Salary: ${average_score:,.2f}")
```
