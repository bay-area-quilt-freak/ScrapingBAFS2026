# We demand `txt`

We'll want to convert things like PDFs and Word files into plain text before we do stuff with them
in Python. This Jupyter Notebook will explore a couple of the ways we can do that using Python.

We'll use `pymupdf` to convert `pdf`s (and other things like `docx`) to text. First, we'll install
`pymupdf` with our usual

```uv add pymupdf```

From there its reletively straightforward. `pymupdf` cares about pages in our original document so
we'll have to use a loop to ensure we get all of the pages converted.

````python
import pymupdf

document = pymupdf.open("2026_March 11 6:30 pm_Common Council CommitteeDeliberation Session.pdf")

my_pages = []
for page in document:
    my_pages.append(page.get_text().lower())

text = "\n".join(my_pages)
print(text)
````

We could then dump this into a file with our `with open(...)...:` syntax or whatever. We can get
all of the pdf files in a directory

````python
document = pymupdf.open("2026_March 25 6:30 pm_Common Council Regular Session.pdf")

my_pages = []
for page in document:
    my_pages.append(page.get_text().lower())

text = "\n".join(my_pages)

# a list of keywords related to surveillance
flock_keywords = ["flock", "axon", "alpr"]

flock_mentioned = False
for keyword in flock_keywords:
    if keyword in text:
        flock_mentioned = True

print(flock_mentioned)
````

````python
from pathlib import Path

pdfs = list(Path("../03_scraping/bloomington_city_council_agendas").glob("*.pdf"))

for pdf in pdfs:
    try:
        document = pymupdf.open(pdf)
    except:
        continue

    my_pages = []
    for page in document:
        my_pages.append(page.get_text().lower())  # type: ignore

    text = "\n".join(my_pages)

    # a list of keywords related to surveillance
    flock_keywords = ["flock", "axon", "alpr"]

    flock_mentioned = False
    for keyword in flock_keywords:
        if keyword in text:
            flock_mentioned = True

    if flock_mentioned:
        print(pdf)
````

````python
import pandas as pd
def alpr_filter(text, date):

    flock_keywords = ["flock", "axon", "alpr"]

    flock_mentioned = False
    for keyword in flock_keywords:
        if keyword in text and date > pd.to_datetime("2022-01-01"):
            flock_mentioned = True

    return flock_mentioned

def police_filter(text):

    police_keywords = ["police", "bpd", "sergeant", "chief"]

    police_mentioned = False
    for keyword in police_keywords:
        if keyword in text:
            police_mentioned = True

    return police_mentioned

def extract_date(path):
    date_string = " ".join(path.stem.split("_")[:2])
    dt = pd.to_datetime(date_string)
    return dt


pdfs = list(Path("../03_scraping/bloomington_city_council_agendas").glob("*.pdf"))

my_pages = []
for pdf in pdfs:
    document = pymupdf.open(pdf)

    my_pages = []
    for page in document:
        my_pages.append(page.get_text().lower())

    text = "\n".join(my_pages)

    pdf_date = extract_date(pdf)

    if alpr_filter(text, pdf_date):
        print(f"flock mentioned in {pdf}")
    if police_filter(text):
        print(f"police mentioned in {pdf}")
````
