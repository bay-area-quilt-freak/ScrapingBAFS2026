# The DataFrame

Basically 100% of the time we're doing data analysis, we're working with
DataFrames. The `pandas` library provides the DataFrame object and basically
enables all data work in Python. DataFrames are pretty straightforwardly tables
in the sort of spreadsheet sense. `pandas` is aware of what a 'column' is and
provides a lot of ways to filter and clean data.

Just like numpy you may need to install pandas. Use `ctrl` + `~` to open the command terminal. Type `source .venv/bin/activate` hit `Enter` then type `python3 -m pip install pandas` hit `Enter`. Now numpy is installed in your virtual environment!

First, I'll show how we make one from scratch so we have an easy dataframe to
explore functionality with.

```python
import pandas as pd
import numpy as np

# dataframes are amazing
category = ["vowel", "consonant", "consonant", "consonant", "vowel", "consonant", "consonant", "consonant"]
my_df = pd.DataFrame({"labels": ["a", "b", "c", "d", "e", "f", "g", "h"],
                      "numbers": np.random.normal(size=8),
                      "category": category})
print("This is my DataFrame:")
print(my_df)
```

We can add new columns to our dataframe by just pretending it exists and
assigning a value to it. The value MUST be the same length as the other columns!

```python
my_df["big_numbers"] = np.random.normal(loc=10, size=8)
my_df
```

Often, we will want to know some basic info about our dataframe. Two pieces
of information that are really helpful are the `.info()` method, and the .`shape`
attribute

```python
my_df.shape
```

This shows us the number of rows then the number of columns. Sometimes we need
to use these numbers in code we're writing and this allows us to access them
programmatically.

```python
my_df.info()
```

`.info()` is more helpful for when we want to visually inspect the overall
picture of our data. It first lists the number of rows (`RangeIndex:`), then the
number of columns `Data Columns`. Then it tells us each column's name and the
number of real values in that column and the column's type. Missing or *null*
values will be a problem for us when we get our hands on real data.

### Selecting rows and columns

Because dataframes are aware that we want to think about our data as tables, `pandas` has made it incredibly easy to select a column

```python
# we can kind of think of it a little bit like a dictionary of numpy arrays
# we call the the key-value pair a column
print("My 'labels' column is:")
print(my_df["labels"])
```

We can even select multiple columns at the same time using a list of column
names rather than a single column name.

```python
my_df[["labels", "category"]]
```

The numbers on the left next to our column values are called the Index. We can
use the Index to select rows rather than columns. We will use `iloc` to select
rows using the Index (if the Index is an integer as it is here). In cases where
our Index is a string, which might come up occasionally, we will use `loc`
instead.

> side note: the Index can be very powerful. `pandas` gives us lots of ways of
> doing things and I typically avoid using the Index when possible. some people
> love it

```python
my_df.iloc[2]
```

The format looks a little different (it prints the row vertically like a column)
but if you check against the table you should find that it matches with the
table we printed out earlier.

We can use the Index along with a column name to select individual values

```python
my_df.iloc[2]["category"]
```

```python
# Now try using the index to locate the row f and select the "big numbers" value.
my_df.iloc[5]["big_numbers"]
```

### Element-wise operations

One of the core features of `pandas` is the ability to do element-wise
operations on columns, much like numpy arrays. This means we can write a single statement and without
looping, do the operation on each row of the dataframe column

```python
# this will double each number in the "numbers" column|
print(my_df["numbers"])

print(my_df["numbers"] * 2)
```

We can do something similar using two columns rather than one column and a
literal value.

```python
my_df["labels"] + my_df["category"]
```

We can even combine element-wise operations with helpful built-in column methods.

For example we can rescale a column based on the maximum value in a single line of code.

```python
my_df["numbers"] / my_df["numbers"].max()
```

Not helpful in this case but still pretty cool to be able to do that. It will be
much more useful when we see dataframes with more interesting columns.

For reasons that will hopefully soon be very obvious, the most important
element-wise operations we can do are conditional statements. 

Lets create a mask variable. 
Instead of merely printing the result of our element-wise operation, lets also assign it to a
variable. We'll use it a little later.

```python
positive_numbers = my_df["numbers"] > 0
positive_numbers
```

We get a column of boolean values when we do this. We can use this to *filter* our data.

### Filtering data

dataframes can also use so-called fancy indexing that we learned about in numpy arrays. In most cases, we will do
fancy indexing with boolean arrays like the one we've just constructed. When we
use a boolean array to index, we call it *masking* with the boolean array being
called a *mask*.

```python
# we can use fancy indexing to pull out specific rows
my_df[positive_numbers].reset_index()
```

Unless we're using a complicated set of conditionals with multiple `and`s and `or`s we usually don't bother giving the mask its own variable (but feel free if it helps you). Instead we'll just put the conditional directly in the square braces used to call the index.

```python
my_df[my_df["numbers"] > 0]
```

In doing this fancy indexing, we've pulled out just the rows of the dataframe
for which the numbers column is positive. Its really helpful to be able to
perform this kind of filtering so easily. If we wanted to do this using lists
everytime we wanted to filter we would die before we ever got to anything cool.


Often we will want to pull out a single column after filtering. We can do this
by using the same syntax we used to select a column stuck on the end of our
fancy indexing with additional square braces.

```python
my_df[my_df["numbers"] > 0]["labels"]
```
