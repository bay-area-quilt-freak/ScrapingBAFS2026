## For Loops
Now that you know about lists we can talk about for loops!! 
For loops are crazy useful and we can abuse them like crazy! 

Lets say each item in that list you want to print it out on it's 
own and also print out 'YAY!' each time, well you are in LUCK. 
Lets iterate through each value in that list using a for loop.

```python
# Note the colon, it marks the beginning of what goes INSIDE your for loop
for x in string_list:
    print(x) #  everything indented after the for loop line is INSIDE the for loop.
print('YAY!')# this line is not inside the for loop....
```

hmmmm that didn't work. It only printed 'YAY!' once at the end... I wanted it to print every time I printed a list item. 

We need to move the `print('YAY!')` statment in to the for loop. INDENTATION MATTERS!!!!!

```python
# ok so what this for loop is doing is taking each value in the string_list and assigning it to the variable 'x'. 
for x in string_list:
    print(x)
    print('YAY!')
```

```python
# you can use any varible name in the for loop, it doesn't have to be x!
for beers in my_list:
    print(beers)
    print('beers on the wall!')
```

## Indexing
SO.... you know about lists and you know how to do something for every item in
 your list using a for loop but what if you just want one item from your list?
Lets's say I want to know what the 1st element in my_list is... we can call the 
element with it's index number!

We access elements in our list by
indexing with integer values that correspond to the positions of the list
elements. Like most other programming languages, indexing in python starts at 0. 


| index | value |
| ----- | ----- |
| 0 | 420 |
| 1 | 6 |
| 2 | 2 |
| 3 | 8 |
| 4 | 3000 |



```python
# to index our list we use square brakets.
#lets print out the first element in my list.
print(my_list[0])
```

### Now you try!

We can also grab a **slice** of a list. This is a contigious chunk of elements
in our list. Still inside the indexing square brackets but now we use the syntax

`[start_index (inclusive):end_index (exclusive)]`

Here by inclusive and
exclusive I mean that the element at index `start_index` will be in your slice
but the element at `end_index` will not. 

**We will see this over and over again.**

```python
print(my_list[0:3])

#We can also get slices without specifying a start or end part and grab the beginning or end of the list
print(my_list[:3])
```

Ok now this is gonna be CRAZY but we are going to combine our knowledge on indexing and for loops and use the function 'ennumerate'.
enumerate is really useful if we want to know the index number and the value of each item in our list. 
## Enumerate with For loops

```python
yet_another_list = ['filth', 'taudry', 'trashy', 'hot']
# note the colon in the following line!! 
for i, value in enumerate(yet_another_list): 
    print(i) # this is the index
    print(value) # this is the value at index i in the list
```

```python
# we again can use any variable name for the index and list element 
for indx, beep in enumerate(yet_another_list):

    # this is some cute formating! Note the f"" and the curly brakets
    # around the variable
    print(f"At index {indx} the value of the list is {beep}")
```
