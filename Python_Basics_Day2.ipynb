## For Loops
Now that you know about lists we can talk about for loops!! 
For loops are crazy useful and we can abuse them like crazy! 

Lets say that you want to print out each item in that list on it's own
and also print out 'YAY!' each time, well you are in LUCK. 
Lets iterate through each value in that list using a for loop.

````python
# Note the colon, it marks the beginning of what goes INSIDE your for loop
for x in string_list:
    print(x) #  everything indented after the for loop line is INSIDE the for loop.
print('YAY!')# this line is not inside the for loop....
````

hmmmm that didn't work. It only printed 'YAY!' once at the end... I wanted it to print every time I printed a list item.

````python
# ok so what this for loop is doing is taking each value in the string_list and assigning it to the variable 'x'. 
for x in string_list:
    print(x)
    print('YAY!')
print(x)
````

Numpy arrays can only contain values of a single type. Integers (7) get
converted to floats (7.0) if you try to combine them in a single array.
Everything gets converted to a string if you try to combine them with strings.

Technically the number of elements in an array cannot change
once the array is created. Though you will find that we can make copies of arrays quite easily to manipulate them.

We can edit lists then convert them into numpy arrays once we have the values we want.

````python
import numpy as np

# you can make a numpy array out of a list
my_list = [1, 2, 3.0, 4, 5.134]
my_array = np.array(my_list)

my_string_array = np.array(["1", 2, 3.0, 4, 5.134])

print(my_array)
print(my_string_array)
````

numpy arrays are indexed the same way as lists.
You can look at single values as well as slices.

````python
print(my_list[2])
print(my_string_array[1:])
````

As the name suggests, numpy is for working with numbers. The array is generally
how we do it. Numpy comes with a ton of built in math functionality.

````python
# some basic functionality
print("the mean:")
print(np.mean(my_array))
print('the median:')
print(np.median(my_array))
````

Numpy arrays themselves support "element-wise" operations on their contents.
This let's us avoid loops in many cases which makes our code work much better.
We can iterate over arrays like we do lists, and make numerical changes to the
values as we would any other number.

````python
# note that you need to create the array you will iterate through before the for loop. 
# This is because technically the size of the array cannot be changed.
even_number_array = np.array([2, 4, 6, 8, 10])

# using one of the many sick af numpy array creation routines!
# np.zeros_like() creates an array full of zeros that is the 
# same shape as the array you give it!
loop_odd_array = np.zeros_like(even_number_array)

for i, x in enumerate(even_number_array):
    loop_odd_array[i] = x + 1

print(even_number_array)
print(loop_odd_array)
````

However, we can simply *add one to the array* with numpy and get the same
result. We can even add two arrays together to add corresponding elements. The
same goes for any binary mathematical operation.

````python
faster_odd_array = even_number_array + 1
print(faster_odd_array)

print(faster_odd_array+even_number_array)
````

Look at the last two blocks of code. They both do the same thing but we were 
able to do everything that for loop did in one single line of code. 

## Functions

Functions are code that we want to use over and over again. You've already seen
them, `print()`, `type()`, and `.endswith` are all functions. Python has a lot of
useful functions built in and we will use *libraries* to provide us
with additional functions.


In order to understand how functions work, we're going to make our own.

````python

def add_two_numbers(a, b):
    total = a + b
    return total

k = add_two_numbers(3, 4)
````

Let's unpack this. The first line is called the function signature. It contains
the name of the function as well as what *arguments* it takes. Our
`add_two_numbers` function takes two arguments, `a` and `b`. The names of the
arguments are how we refer to them INSIDE the definition of the function. They are
like aliases we use to define the logic inside the function, when we *call* the
function, the first thing we put in the parentheses is "a" and the second is
"b". We can also be explicit about which variable is which regardless of order
if we want to (`add_two_numbers(b=4, a=3)`)

Inside of the function, we're defining a variable, `total` that contains the sum
of `a` and `b`. `return` tells python that we want the function to evaluate to
whatever comes after it when called. In the example above, `add_two_numbers(3,
4)` evaluates to the internal variable we called `total` which is, of course, 7.
This allows us to assign the result of the function to a variable and so on.


````python
first_number = 6
second_number = 7

first_plus_second = add_two_numbers(first_number, second_number)

print(first_plus_second)
````

There are many resons we will want to create our own fuctions but the two main ones are for orginization and to be able to repeatedly call a slightly complex evaluation (in a function). To understand why functions help keep your code organized we need to understand global variables and local variables. In the case above the variables `first_number` and `second_number` are global varibles and the variables `a` and `b` are local varaibles that can only be called in our function `add_two_numbers`. A global varible is any varible defined outside of a function and can be called any where in python script. We will generally want to avoid using global variables, though they can be usefull for constants like pi or what have you.

To avoid using global variables we can call functions within other functions. The standard practice is to write the main body of your script in a function named `main()`. 
Note below that we call two different functions in the function `main`. Also note that we call `main()` at the end of the cell.
