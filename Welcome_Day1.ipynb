### Jupyter Notebooks 
When we run a Python script, the Python interpreter goes 
line-by-line executing each statement and loading things into
memory for the duration of the script.

In the context of data science, we could very well write a
Python script that scrapes a dataset off a webpage, performs some statistical operation, 
and then produces a plot. Sometimes that's exactly what we want to do. But what if all we want to do is
fiddle around with our scraped data and mess with how the formating look? What if
scraping the data takes 5 minutes, and doing the formating takes a couple of seconds?

Lucky for us someone has solved this problem.

Throughout this course we'll be using a tool called *Jupyter Notebooks* to run little small blocks of
code while keeping varibales and libraries in memory. This allows us to isolate the steps of our
overall analysis so we can get away with running the long parts *only once* but running the parts that
require a lot of hands-on fiddling over and over again.
