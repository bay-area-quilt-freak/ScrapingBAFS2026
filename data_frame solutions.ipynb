```python
import pandas as pd
import numpy as np
# only good cop is a dead cop

# We have all our data as lists currently. Use them to create a data frame.

name = ["James", "Daniel", "Matthew", "David", "Christopher", "Andrew", "Sarah", "Thomas", "Michael", "Mark", "Robert", "Richard"]

city = ["Minneapolis", "Oakland", "New York", "Los Angelos", "Los Angelos", "New York", "New York", "Atlanta", "Minneapolis", "Atlanta", "Atlanta", "Chicago"]

cause_of_death = ["Car Crash", "Exploded", "Friendly Fire", "Friendly Fire", "Heart Attack", "Heart Attack", "Decapitated", "Friendly Fire", "Exploded", "Exploded", "Touching Fentanyl", "Shoot Out"]

number_filed_complaints = [203, 3855, 476, 1052, 1312, 327, 7068, 935, 287, 493, 5328, 855]

marriage_status = [ "Divorced", "Single", "Divorced", "Divorced", "Single", "Divorced", "Divorced", "Divorced", "Divorced", "Divorced", "Divorced", "Divorced"]

Blood_Alcohol_Concentration_at_TOD = [0.11, 0.00, 0.04, 0.23, 0.18, 0.01, 0.09, 0.00, 0.08, 0.00, 0.29, 0.35]

rank = ["Officer", "Officer", "Lieutenant", "Lieutenant", "Sergeant", "Officer", "Sergeant", "Lieutenant", "Sergeant", "Lieutenant", "Sergeant", "Officer"]

annual_salary = [97742, 106000, 160884, 153163, 143117, 60532, 93273, 70540, 126410, 98320, 84732, 91400] 

annual_overtime_pay = [150670, 490000, 102050, 98040, 170900, 32700, 27005, 30090, 40070, 26790, 43040, 25700]

# create a data frame here
dead_cop_df = pd.DataFrame({"names": name,
                      "cities": city,
                      "death": cause_of_death,
                      "complaints": number_filed_complaints,
                      "marriage": marriage_status,
                      "BAC": Blood_Alcohol_Concentration_at_TOD,
                      "rank": rank,
                      "salary": annual_salary,
                      "overtime": annual_overtime_pay})
```

Now that you have a data frame, lets figure out how the cop with the largest annual salary died, print out his name, salary, and cause of death. 


Then calculate the effective annual salary by combing the overtime pay and salary and find the cause of death of the cop with the largest effective annual salary. Print out their name, efective salary, and cause of death.

```python
print("Cop with the largest salary:")
print(dead_cop_df[dead_cop_df["salary"] == dead_cop_df["salary"].max()][["names", "death", "salary"]])

dead_cop_df["effective salary"] = dead_cop_df["salary"]+dead_cop_df["overtime"]

print("cop with the largest effective salary:")
print(dead_cop_df[dead_cop_df["effective salary"] == dead_cop_df["effective salary"].max()][["names", "death", "salary"]])
```

We can create a more complicated mask to figure out how much money was saved nationally by 
the death of divorced cops that died by friendly fire or exploding. We will need to created a more
complicated mask in order to do this!

```python
frendfire_exploded_mask = (dead_cop_df["death"] == "Eploded") |  (dead_cop_df["death"] == "Friendly Fire")

divorce_frenfire_exploded_mask = (dead_cop_df["marriage"] == "Divorced") & (frendfire_exploded_mask)



# got some fancy formating to make it have the comma. Adding
# :,.2f
# within the {} brackets at the end formats it that way. 
print(f"The nation saved ${dead_cop_df[divorce_frenfire_exploded_mask]['effective salary'].sum():,.2f} this year!")
```
